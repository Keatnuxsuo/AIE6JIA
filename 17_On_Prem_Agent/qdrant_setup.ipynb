```python
from langchain_ollama import OllamaEmbeddings

embeddings = OllamaEmbeddings(
    model="mxbai-embed-large",
)
```

```python
from langchain_community.document_loaders import JSONLoader
from langchain_community.document_loaders import DirectoryLoader

json_loader = DirectoryLoader(
    path="./data/data",
    glob="**/*.json",
    loader_cls=JSONLoader,
    loader_kwargs={"jq_schema" : "..", "text_content" : False}
)
json_documents = json_loader.load()
```

```python
len(json_documents)
```

> NOTE: This cell will take a while (~15min) as there are a lot of documents - please ensure you'd selected your own data before running this cell.

```python
from langchain_qdrant import QdrantVectorStore

url = "127.0.0.1:6334"
qdrant = QdrantVectorStore.from_documents(
    json_documents,
    embeddings,
    url=url,
    prefer_grpc=True,
    collection_name="DnD_Documents",
)
```

```python
from qdrant_client import QdrantClient

client = QdrantClient("127.0.0.1", port=6333)  # HTTP port
print(client.get_collections())  # List all collections

# To get info about your collection:
print(client.get_collection("DnD_Documents"))
```

```python
# Scroll through points in the collection
result = client.scroll(
    collection_name="DnD_Documents",
    limit=10,
)

for point in result[0]:
    print(point)
```
